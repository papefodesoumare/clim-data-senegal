```python
import pandas as pd 
import numpy as np 
import random # pour choisir un élément au hasard dans une liste 
```

```python
# On "verrouille" le hasard de NumPy sur la valeur 42
#  si on relance ce code , on obtient EXACTEMENT les mêmes nombres aléatoires
np.random.seed(42)

# Pareil, mais pour le module random les deux ont besoin d'être verrouillés séparément
random.seed(42)
```

```python
# On liste les 14 régions  du Sénégal
regions = ["Dakar", "Thies", "Diourbel", "Fatick", "Kaolack", 
           "Kaffrine", "Kolda", "Sedhiou", "Ziguinchor", 
           "Louga", "Saint-Louis", "Matam", "Tambacounda", "Kedougou"]
```

```python
regions
```

```python
# On  crée 500 ménages fictifs
nb_menages = 500

# On liste les cultures possibles
cultures = ["Mil", "Arachide", "Riz", "Maïs", "Sorgho"]

# On crée une liste vide qui va accueillir chaque ménage un par un
liste_menages = []

# On répète cette boucle 500 fois (une fois par ménage)
for i in range(1, nb_menages + 1):

    # On construit l'identifiant du ménage par exemple "MEN00001"
    id_menage = "MEN" + str(i).zfill(5)

    # On choisit une région au hasard parmi les 14
    region = random.choice(regions)

    # On tire un nombre de personnes au hasard entre 3 et 14
    taille_menage = random.randint(3, 14)

    # On tire une superficie cultivée au hasard entre 0.5 et 10 hectares
    superficie_cultivee_ha = round(random.uniform(0.5, 10), 1)

    # On tire un revenu agricole au hasard entre 100 000 et 2 000 000 FCFA
    revenu_agricole_fcfa = random.randint(100000, 2000000)

    # On tire au hasard 0 (pas d'accès à l'eau) ou 1 (accès à l'eau)
    acces_eau_potable = random.randint(0, 1)

    # On choisit une culture principale au hasard
    culture_principale = random.choice(cultures)

    # On tire au hasard 0 (irrigué) ou 1 (dépend de la pluie)
    depend_pluie = random.randint(0, 1)

    # On rassemble toutes les infos de CE ménage dans un petit dictionnaire
    un_menage = {
        "id_menage": id_menage,
        "region": region,
        "taille_menage": taille_menage,
        "superficie_cultivee_ha": superficie_cultivee_ha,
        "revenu_agricole_fcfa": revenu_agricole_fcfa,
        "acces_eau_potable": acces_eau_potable,
        "culture_principale": culture_principale,
        "depend_pluie": depend_pluie,
    }

    # On ajoute ce ménage à notre grande liste
    liste_menages.append(un_menage)

# Une fois les 500 ménages créés, on transforme toute la liste en tableau Pandas
df_menages = pd.DataFrame(liste_menages)

# On affiche les 5 premières lignes pour vérifier que tout a l'air correct
df_menages.head()
```

```python
# On sauvegarde le tableau df_menages dans un fichier CSV
df_menages.to_csv("../data/raw/menages_fictifs.csv", index=False)

# On confirme que ça a marché
print("Fichier sauvegardé avec succès !")
```

```python
# On liste les années et les mois qu'on va couvrir
annees = [2020, 2021, 2022, 2023, 2024]
mois = list(range(1, 13))  # de 1 à 12

# On liste les mois de saison des pluies au Sénégal (juin à octobre, grosso modo)
mois_pluvieux = [6, 7, 8, 9, 10]

# Liste vide qui va accueillir chaque ligne climatique
liste_climat = []

# On répète pour chaque région
for region in regions:
    # On répète pour chaque année
    for annee in annees:
        # On répète pour chaque mois
        for m in mois:

            # Si on est dans un mois de saison des pluies, on tire une pluviométrie plus forte
            if m in mois_pluvieux:
                precipitation_mm = round(random.uniform(50, 300), 1)
            else:
                precipitation_mm = round(random.uniform(0, 20), 1)

            # La température varie un peu selon le mois (plus chaud en saison sèche chaude, avril-mai)
            if m in [4, 5]:
                temperature_moy_c = round(random.uniform(30, 38), 1)
            else:
                temperature_moy_c = round(random.uniform(24, 32), 1)

            # On rassemble les infos de cette ligne climatique
            une_ligne_climat = {
                "region": region,
                "annee": annee,
                "mois": m,
                "precipitation_mm": precipitation_mm,
                "temperature_moy_c": temperature_moy_c,
            }

            # On ajoute cette ligne à la grande liste
            liste_climat.append(une_ligne_climat)

# On transforme toute la liste en tableau Pandas
df_climat = pd.DataFrame(liste_climat)

# On affiche les 5 premières lignes pour vérifier
df_climat.head()
```

```python
# On sauvegarde le tableau df_climat dans un fichier CSV
df_climat.to_csv("../data/raw/climat_fictif.csv", index=False)

# On confirme
print("Fichier climat sauvegardé avec succès !")
```

```python
# On réutilise les mêmes années et mois que pour le climat
# Liste vide qui va accueillir chaque ligne NDVI
liste_ndvi = []

# On répète pour chaque région
for region in regions:
    # On répète pour chaque année
    for annee in annees:
        # On répète pour chaque mois
        for m in mois:

            # En saison des pluies, la végétation est plus verte (NDVI plus élevé)
            if m in mois_pluvieux:
                ndvi = round(random.uniform(0.4, 0.8), 2)
            else:
                ndvi = round(random.uniform(0.1, 0.4), 2)

            # On rassemble les infos de cette ligne NDVI
            une_ligne_ndvi = {
                "region": region,
                "annee": annee,
                "mois": m,
                "ndvi": ndvi,
            }

            liste_ndvi.append(une_ligne_ndvi)

# On transforme la liste en tableau Pandas
df_ndvi = pd.DataFrame(liste_ndvi)

# On affiche les 5 premières lignes
df_ndvi.head()
```

```python
# On sauvegarde le tableau df_ndv dans un fichier CSV
df_ndvi.to_csv("../data/raw/ndvi_fictif.csv", index=False)

# On confirme
print("Fichier indice de vegetation sauvegardé avec succès !")
```

```python
# On calcule la précipitation moyenne et la température moyenne, par région
# .groupby("region") = on regroupe toutes les lignes qui ont la même région ensemble
# .mean() = on calcule la moyenne des colonnes numériques pour chaque groupe
climat_moyen_region = df_climat.groupby("region")[["precipitation_mm", "temperature_moy_c"]].mean().reset_index()

# On affiche pour vérifier
climat_moyen_region
```

```python
# On calcule le NDVI moyen par région, sur les 5 années
ndvi_moyen_region = df_ndvi.groupby("region")[["ndvi"]].mean().reset_index()

# On affiche pour vérifier
ndvi_moyen_region
```

```python
# On fusionne df_menages avec climat_moyen_region, en reliant sur la colonne "region"
df_final = df_menages.merge(climat_moyen_region, on="region", how="left")

# On fusionne le résultat avec ndvi_moyen_region, toujours sur "region"
df_final = df_final.merge(ndvi_moyen_region, on="region", how="left")

# On affiche les 5 premières lignes pour vérifier
df_final.head()
```

```python

# Une petite fonction qui ramène n'importe quelle colonne entre 0 et 1
def normaliser(colonne):
    return (colonne - colonne.min()) / (colonne.max() - colonne.min())

# On normalise chaque colonne qui nous intéresse
df_final["precipitation_norm"] = normaliser(df_final["precipitation_mm"])
df_final["ndvi_norm"] = normaliser(df_final["ndvi"])
df_final["revenu_norm"] = normaliser(df_final["revenu_agricole_fcfa"])
df_final["superficie_norm"] = normaliser(df_final["superficie_cultivee_ha"])

# Vérification 
df_final[["precipitation_norm", "ndvi_norm", "revenu_norm", "superficie_norm"]].describe()
```

```python
# On calcule le score d'EXPOSITION
# Plus la précipitation est basse ET le NDVI est bas, plus l'exposition est forte
# (1 - precipitation_norm) inverse l'échelle : une faible pluie donne un score élevé
score_exposition = (1 - df_final["precipitation_norm"]) * 0.5 + (1 - df_final["ndvi_norm"]) * 0.5

# On calcule le score de SENSIBILITÉ
# depend_pluie vaut déjà 0 ou 1, donc pas besoin de le normaliser
# (1 - superficie_norm) : une petite superficie = plus sensible
score_sensibilite = df_final["depend_pluie"] * 0.5 + (1 - df_final["superficie_norm"]) * 0.5

# On calcule le score de CAPACITÉ D'ADAPTATION
# acces_eau_potable vaut déjà 0 ou 1
score_adaptation = df_final["acces_eau_potable"] * 0.5 + df_final["revenu_norm"] * 0.5

# On combine tout : exposition + sensibilité, MOINS la capacité d'adaptation
# (une bonne capacité d'adaptation RÉDUIT la vulnérabilité)
df_final["indice_vulnerabilite"] = (score_exposition * 0.4 + score_sensibilite * 0.3 
                                      + (1 - score_adaptation) * 0.3)

# On arrondit à 2 décimales pour que ce soit plus lisible
df_final["indice_vulnerabilite"] = round(df_final["indice_vulnerabilite"], 2)

# On affiche les ménages les plus vulnérables en premier
df_final.sort_values("indice_vulnerabilite", ascending=False)[["id_menage", "region", "indice_vulnerabilite"]].head(10)
```

```python
# On sauvegarde le tableau final avec l'indice de vulnérabilité dans data/processed
df_final.to_csv("../data/processed/menages_avec_indice_vulnerabilite.csv", index=False)

print("Fichier final sauvegardé avec succès !")
```

```python
# On calcule l'indice de vulnérabilité moyen par région
vulnerabilite_par_region = df_final.groupby("region")[["indice_vulnerabilite"]].mean().reset_index()

# On arrondit pour que ce soit plus lisible
vulnerabilite_par_region["indice_vulnerabilite"] = round(vulnerabilite_par_region["indice_vulnerabilite"], 2)

# On trie du plus vulnérable au moins vulnérable
vulnerabilite_par_region = vulnerabilite_par_region.sort_values("indice_vulnerabilite", ascending=False)

vulnerabilite_par_region
```

```python
# Coordonnées approximatives (latitude, longitude)  de chaque région
coordonnees_regions = {
    "Dakar": (14.6928, -17.4467),  
    "Thies": (14.7833, -16.9167),
    "Diourbel": (14.6521, -16.2340),
    "Fatick": (14.3390, -16.4110),
    "Kaolack": (14.0167, -16.2500),      
    "Kaffrine": (14.1059, -15.5508),
    "Kolda": (12.8833, -14.9500),
    "Sedhiou": (12.7081, -15.5569),
    "Ziguinchor": (12.5833, -16.2719),
    "Louga": (15.6173, -16.2240),
    "Saint-Louis": (16.0179, -16.4896),
    "Matam": (15.6170, -13.3330),
    "Tambacounda": (13.7707, -13.6673),
    "Kedougou": (12.5605, -12.1747),
}

# On ajoute latitude et longitude au tableau vulnerabilite_par_region
vulnerabilite_par_region["latitude"] = vulnerabilite_par_region["region"].map(lambda r: coordonnees_regions[r][0])
vulnerabilite_par_region["longitude"] = vulnerabilite_par_region["region"].map(lambda r: coordonnees_regions[r][1])

vulnerabilite_par_region
```

```python
import folium
# On utilise Esri comme fournisseur de fond de carte (gratuit, sans clé API)
carte = folium.Map(
    location=[14.5, -14.5],
    zoom_start=7,
    tiles="https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}",
    attr="Esri"
)

# Une petite fonction qui choisit une couleur selon le niveau de vulnérabilité
def choisir_couleur(indice):
    if indice >= 0.6:
        return "red"      # très vulnérable
    elif indice >= 0.45:
        return "orange"   # moyennement vulnérable
    else:
        return "green"    # peu vulnérable

# On parcourt chaque région du tableau vulnerabilite_par_region
for index, ligne in vulnerabilite_par_region.iterrows():

    # On choisit la couleur du marqueur selon son indice
    couleur = choisir_couleur(ligne["indice_vulnerabilite"])

    # On crée le texte qui s'affichera au clic sur le marqueur
    texte_popup = f"{ligne['region']} : indice = {ligne['indice_vulnerabilite']}"

    # On ajoute un marqueur circulaire sur la carte
    folium.CircleMarker(
        location=[ligne["latitude"], ligne["longitude"]],
        radius=12,
        popup=texte_popup,
        color=couleur,
        fill=True,
        fill_color=couleur,
        fill_opacity=0.7,
    ).add_to(carte)

# On affiche la carte  dans le notebook
carte
```

```python
# On sauvegarde la carte dans un fichier HTML autonome
carte.save("../outputs/carte_vulnerabilite.html")

print("Carte sauvegardée avec succès !")
```
